```python
# [1주차 실습] 파이쎤 기본 4대 자료구조 실습



# 1. 리스트 (list) : 순서가 있고 수정 가능한 가변 자료구조
my_list = [10, 20, 30, 'AI', 'Video']
print("리스트 데이터:", my_list)
print("데이터 타입:", type(my_list))
print("-" * 30)

# 2. 튜플 (Tuple)
my_tuple = (100, 200, 'processing')
print("=== 2. Tuple 실습 ===")
print("튜플 데이터:", my_tuple)
print("데이터 타입:", type(my_tuple))
print("-" * 30)

# 3. 셋 (set)
my_set = {1, 2, 3, 3, 2, 1, 'Data'}
print("=== 3. set 실습 (중복 제거 확인) ===")
print("셋 데이터:", my_set)
print("데이터 타입:", type(my_set))
print("-" * 30)

# 4. 딕셔너리 (Dictionary)
my_dict = {
    'subject': 'AI Video Data Processing',
    'week' : 1,
    'major': 'Business Administration',
    'status': 'completed'
}
print("=== 4. Dictionary 실습 ===")
print("딕셔너리 데이터:", my_dict)
print("데이터 타입:", type(my_dict))
```
