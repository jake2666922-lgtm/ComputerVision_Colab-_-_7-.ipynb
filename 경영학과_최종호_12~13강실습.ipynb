```python
from pathlib import Path
import cv2
from google.colab.patches import cv2_imshow

#코랩 기본 작업 경로(/content) 기준 이미지 로드
parrot_file = Path("/content/왕관앵무새.jpeg")
parrot_bgr = cv2.imread(str(parrot_file))

assert parrot_bgr is not None, f"파일 로드 실패: {parrot_file}"

# 밝기 채널만 추출하기 위해 그레이스케일 변환
parrot_mono = cv2.cvtColor(parrot_bgr, cv2.COLOR_BGR2GRAY)

#검출기 인스턴스 생성 (SIFT: 상위 30개 제한, ORB: 상위 20개 제한)
sift_finder = cv2.SIFT_create(nfeatures=30)
orb_finder = cv2.ORB_create(nfeatures=20)

# 각 알고리즘별 키포인트와 디스크립터 산출
sift_pts, sift_desc = sift_finder.detectAndCompute(parrot_mono, None)
orb_pts, orb_desc = orb_finder.detectAndCompute(parrot_mono, None)

# 검출된 특징점 수 요약 출력
print(f"[SIFT] 검출 수: {len(sift_pts)}개 | [ORB] 검출 수: {len(orb_pts)}개")

# 시각화 이미지 생성 (색상 및 플래그 차별화)
vis_sift = cv2.drawKeypoints(
    parrot_bgr,
    sift_pts,
    None,
    color=(0, 255, 0),
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS,
)
vis_orb = cv2.drawKeypoints(
    parrot_bgr,
    orb_pts,
    None,
    color=(0, 165, 255),
    flags=cv2.DRAW_MATCHES_FLAGS_DEFAULT,
)

# 코랩 화면 출력
print("--- SIFT 특징점 ---")
cv2_imshow(vis_sift)

print("--- ORB 특징점 ---")
cv2_imshow(vis_orb)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# 1. 원본 이미지 불러오기 (/content/앵무새여러마리.jpeg 기준)
src_path = "/content/앵무새여러마리.jpeg"
base_img = cv2.imread(src_path)

if base_img is None:
    raise FileNotFoundError(f"이미지를 찾을 수 없습니다: {src_path}")

# 한 장으로 매칭 실습을 하기 위해 회전 + 크기 조절한 변환 이미지 생성
height, width = base_img.shape[:2]
rot_matrix = cv2.getRotationMatrix2D((width / 2, height / 2), angle=-12, scale=0.85)
warped_img = cv2.warpAffine(base_img, rot_matrix, (width, height))

# 2. 그레이스케일 변환
base_gray = cv2.cvtColor(base_img, cv2.COLOR_BGR2GRAY)
warped_gray =cv2.cvtColor(warped_img, cv2.COLOR_BGR2GRAY)

# 3. ORB 특징점 및 디스크립터 추출
orb_model = cv2.ORB_create(nfeatures=500)
kp_base, des_base = orb_model.detectAndCompute(base_gray, None)
kp_warped, des_warped = orb_model.detectAndCompute(warped_gray, None)

# 4. Hamming 거리 기반 매칭 및 거리 기준 필터링
matcher = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
raw_matches = matcher.match(des_base, des_warped)

# 거리가 짧은(유사도가 높은) 순으로 정렬 후 상위 50개만 선택)
good_matches = sorted(raw_matches, key=lambda m: m.distance)[:50]
print(f"전체 매칭 수: {len(raw_matches)}개 > 상위 선택: {len(good_matches)}개")

# 5. 매칭 결과 시각화
matches_view = cv2.drawMatches(
    base_img,
    kp_base,
    warped_img,
    kp_warped,
    good_matches,
    None,
    matchColor=(0, 255, 0),
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS,
)

cv2_imshow(matches_view)
```

```python
import cv2
import matplotlib.pyplot as plt

img_path = "/content/썬코뉴어.jpeg"
img_bgr = cv2.imread(img_path)

if img_bgr is None:
  raise FileNotFoundError(f"이미지를 불러올 수 없습니다: {img_path}")

img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 1. ORB 특징점 검출기 생성 및 검출
orb = cv2.ORB_create(nfeatures=800, scaleFactor=1.2, nlevels=8)
kp1, des1 = orb.detectAndCompute(img_rgb, None)

# 2. 이미지 회전(35도) 및 스케일(0.85배) 변환으로 비교용 테스트 이미지 생성
h, w = img_rgb.shape[:2]
rot_matrix = cv2.getRotationMatrix2D((w / 2, h / 2), angle=35, scale=0.85)
img_transformed = cv2.warpAffine(img_rgb, rot_matrix, (w, h))

kp2, des2 = orb.detectAndCompute(img_transformed, None)

# 3. FlannBassedMatcher (LSH 인덱스)로 매칭
index_params = dict(
    algorithm=6,  # FLANN_INDEX_LSH
    table_number=6,
    key_size=12,
    milti_probe_level=1,
)
search_params = dict(checks=50)
flann = cv2.FlannBasedMatcher(index_params, search_params)

good_matches = []
if des1 is not None and des2 is not None and len(kp1) >= 2 and len(kp2) >= 2:
  knn_matches = flann.knnMatch(des1, des2, k=2)
  for pair in knn_matches:
    if len(pair) == 2:
      m, n = pair
      if m.distance < 0.7 * n.distance:
        good_matches.append(m)

# 4. 결과 사진 만들기 및 화면에 띄우기
result = cv2.drawMatches(
    img_rgb,
    kp1,
    img_transformed,
    kp2,
    good_matches[:30],
    None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS,
)
cv2_imshow(cv2.cvtColor(result, cv2.COLOR_RGB2BGR))
```

```python
import cv2
from google.colab.patches import cv2_imshow

# 1. 금강앵무새 이미지 불러오기
img_bgr = cv2.imread("/content/금강앵무새.jpeg")

if img_bgr is None:
  raise FileNotFoundError(
      "이미지를 불러올 수 없습니다. '/content/금강앵무새.jpeg' 경로를 확인해 주세요."
  )

# 2. 비교용 테스트 이미지 만들기 (사진 속 SURF/FLANN 방식과 완전히 다르게: 좌우 반전 후 45도 회전 및 작게 축소)
img_flip = cv2.flip(img_bgr, 1) # 좌우 반전
h, w = img_flip.shape[:2]
rot_matrix = cv2.getRotationMatrix2D((w / 2, h / 2), angle=45, scale=0.7)
img_transformed = cv2.warpAffine(img_flip, rot_matrix, (w, h))

# 3. 사진 속 알고리즘과 다른 ORB 특징점 검출기 사용
detector = cv2.ORB_create(nfeatures=1000)
kp1, des1 = detector.detectAndCompute(img_bgr, None)
kp2, des2 = detector.detectAndCompute(img_transformed, None)

# 4. 사진 속 FLANN 대신, 완전 다른 매칭 방식(BFMatcher)과 해밍 거리(Hamming Distance) 매칭 사용
# knnMatch와 비율 테스트(Love's Ratio Test) 방식
bf = cv2.BFMatcher(cv2.NORM_HAMMING)

good_matches = []
if des1 is not None and des2 is not None:
  matches = bf.knnMatch(des1, des2, k=2)

  # 거리 비율 테스트로 좋은 매칭만 선별 (m.distance < 0.75 * n.distance)
  for pair in matches:
    if len(pair) == 2:
      m, n = pair
      if m.distance < 0.75 * n.distance:
        good_matches.append(m)

# 5. 매칭 결과 시각화 및 화면 출력 (BGR 색상 온전히 보존)
result_img = cv2.drawMatches(
    img_bgr,
    kp1,
    img_transformed,
    kp2,
    good_matches[:35],  # 상위 35개 좋은 매칭 연결
    None,
    matchColor=(0, 255, 100), # 연결 선 색상 (초록빛)
    singlePointColor=(0, 100, 255), # 특징점 색상 (파란빛)
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS,
)

# 최종 결과 이미지를 화면에 띄우기
cv2_imshow(result_img)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# 1. 뉴기니아앵무새 이미지 불러오기
base_path = "/content/뉴기니아앵무새.jpeg"
src_img = cv2.imread(base_path)

if src_img is None:
  raise FileNotFoundError(f"이미지를 찾을 수 없습니다: {base_path}")

# 2. 비교용 이미지 생성 (상하 반전 후 0.8배 축소)
flipped = cv2.flip(src_img, 0)
rows, cols = flipped.shape[:2]
resized_view = cv2.resize(
    flipped, (int(cols * 0.8), int(rows * 0.8)), interpolation=cv2.INTER_AREA
)

# 3. AKAZE 특징점 검출 및 디스크립터 계산
detector = cv2.ORB_create(nfeatures=1000)
keypoints_a, desc_a = detector.detectAndCompute(src_img, None)
keypoints_b, desc_b = detector.detectAndCompute(resized_view, None)

# 4. crossCheck 기반 브루트포스 매칭 및 거리순 정렬
matcher = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matched_pairs = []

if desc_a is not None and desc_b is not None:
  raw_matches = matcher.match(desc_a, desc_b)
  matched_pairs = sorted(raw_matches, key=lambda item: item.distance)[:25]

# 5. 결과 이미지 그리기 및 화면 출력
output_img = cv2.drawMatches(
    src_img,
    keypoints_a,
    resized_view,
    keypoints_b,
    matched_pairs,
    None,
    matchColor=(255, 100, 0),
    singlePointColor=(0, 200, 255),
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS,
)

cv2_imshow(output_img)
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def warp_pair(img1, img2, H):
  h1, w1 = img1.shape[:2]
  h2, w2 = img2.shape[:2]
  c1 = np.float32([[0, 0], [0, h1], [w1, h1], [w1, 0]]).reshape(-1, 1, 2)
  c2 = cv2.perspectiveTransform(
      np.float32([[0, 0], [0, h2], [w2, h2], [w2, 0]]).reshape(-1, 1, 2), H
  )
  pts = np.vstack((c1, c2))
  xmin, ymin = np.floor(pts.min(axis=0).ravel()).astype(int)
  xmax, ymax = np.ceil(pts.max(axis=0).ravel()).astype(int)
  tx, ty = xmin, -ymin
  T = np.array([[1, 0, tx], [0, 1, ty], [0, 0, 1]], dtype=np.float64)
  out = cv2.warpPerspective(img2, T @ H, (xmax - xmin, ymax - ymin))
  out[ty : ty + h1, tx : tx + w1] = img1
  return out

bgr = cv2.imread("/content/사랑앵무새.jpeg")
if bgr is None:
  raise FileNotFoundError("사랑앵무새.jpeg 없음")

rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
mid = rgb.shape[1] // 2
res = warp_pair(rgb[:, :mid], rgb[:, mid:], np.eye(3))
cv2.imwrite("/content/사랑앵무새_out.png", cv2.cvtColor(res, cv2.COLOR_RGB2BGR))

plt.imshow(res)
plt.axis("off")
plt.show()
```
