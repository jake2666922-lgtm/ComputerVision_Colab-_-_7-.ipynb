```
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt

from skimage import data
src_image = cv2.cvtColor(data.astronaut(), cv2.COLOR_RGB2GRAY)

threshold_val = 130
max_val = 255

ret_bin, img_binary = cv2.threshold(src_image, threshold_val, max_val, cv2.THRESH_BINARY)
ret_inv, img_binary_inv = cv2.threshold(src_image, threshold_val, max_val, cv2.THRESH_BINARY_INV)
ret_trunc, img_trunc = cv2.threshold(src_image, threshold_val, max_val, cv2.THRESH_TRUNC)
ret_tozero, img_tozero = cv2.threshold(src_image, threshold_val, max_val, cv2.THRESH_TOZERO)
ret_tozero_inv, img_tozero_inv = cv2.threshold(src_image, threshold_val, max_val, cv2.THRESH_TOZERO_INV)

processed_images = {
    'ORIGINAL' : np.asarray(src_image, dtype=np.uint8),
    'BINARY' : np.asarray(img_binary, dtype=np.uint8),
    'BINARY_INV' : np.asarray(img_binary_inv, dtype=np.uint8),
    'TRUNC' : np.asarray(img_trunc, dtype=np.uint8),
    'TOZERO' : np.asarray(img_tozero, dtype=np.uint8),
    'TOZERO_INV' : np.asarray(img_tozero_inv, dtype=np.uint8)
}
plt.figure(figsize=(12, 6))
for idx, (title, img_data) in enumerate(processed_images.items()):
        plt.subplot(2, 3, idx + 1)
        plt.imshow(img_data, cmap='gray')
        plt.title(title)
        plt.xticks([])
        plt.yticks([])
plt.tight_layout()
plt.show()
```

```
import cv2
import numpy as np
from skimage import data
import matplotlib.pyplot as plt

img = cv2.cvtColor(data.astronaut(), cv2.COLOR_RGB2GRAY)
img = cv2.resize(img, (300,300))

image1 = img.copy()

image2 = cv2.bitwise_not(img)

maskImage2D = np.zeros((300, 300), dtype=np.uint8)
cv2.circle(maskImage2D, (150, 150), 150, 255, -1)

andImage = cv2.bitwise_and(image1, image2)
orImage = cv2.bitwise_or(image1, image2)
xorImage = cv2.bitwise_xor(image1, image2)
andMask = cv2.bitwise_and(image1, image2, mask=maskImage2D)

processed_images = {
    'IMAGE 1' : image1,
    'IMAGE 2' : image2,
    'AND' : andImage,
    'OR' : orImage,
    'XOR' : xorImage,
    'MASKED AND' : andMask
}
plt.figure(figsize=(12, 8))
for idx, (title, img_data) in enumerate(processed_images.items()):
    plt.imshow(img_data, cmap='gray')
    plt.title(title)
    plt.xticks([])
    plt.yticks([])

plt.tight_layout()
plt.show()

```

```
import cv2
import numpy as np
from skimage import data
from google.colab.patches import cv2_imshow

grayImage = data. camera()
grayImage = cv2.resize(grayImage, (500, 500))

adaptiveMean1 = cv2.adaptiveThreshold(grayImage, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 3, 5)
adaptiveMean2 = cv2.adaptiveThreshold(grayImage, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 5, 5)
adaptiveMean3 = cv2.adaptiveThreshold(grayImage, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 7, 5)
adaptiveMean4 = cv2.adaptiveThreshold(grayImage, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 9, 5)
adaptiveMean5 = cv2.adaptiveThreshold(grayImage, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 5)
adaptiveMean6 = cv2.adaptiveThreshold(grayImage, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 13, 5)

adaptiveMeanImages1 = cv2.hconcat([adaptiveMean1, adaptiveMean2, adaptiveMean3])
adaptiveMeanImages2 = cv2.hconcat([adaptiveMean4, adaptiveMean5, adaptiveMean6])
adaptiveMeanImages = cv2.vconcat([adaptiveMeanImages1, adaptiveMeanImages2])

cv2_imshow(adaptiveMeanImages)
```

```
import cv2
import numpy as np
from matplotlib import pyplot as plt

g80_image = cv2.imread('/G80.jpeg')

g80_multiply_const = 2
g80_divide_const = 2.5

g80_multiply_res = g80_image * g80_multiply_const
g80_divide_res = g80_image / g80_divide_const

print(f"G80 Mltiply Result dtype: {g80_multiply_res.dtype}")
print(f"G80 Divide Resilt dtype: {g80_divide_res.dtype}")

g80_divide_res_uint8 = g80_divide_res.astype('uint8')

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(g80_divide_res_uint8, cv2.COLOR_BGR2RGB))
plt.title(f'divide by {g80_divide_const}')
plt.axis('off')

plt.tight_layout()
plt.show()

print(f"G80 Multiply Res dtype : {g80_multiply_res.dtype}")
print(f"G80 Divide Res dtype: {g80_divide_res.dtype}")

print(f"G80 Final Frame dtype: {g80_divide_res_uint8.dtype}")
```

```
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('G90.jpeg')

hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

blue1 = np.array([100, 50, 50])
blue2 = np.array([140, 255, 255])
red1 = np.array([0, 50, 50])
red2 = np.array([10, 255, 255])
red3 = np.array([170, 50, 50])
red4 = np.array([180, 255, 255])

mask_blue = cv2.inRange(hsv, blue1, blue2)
mask_red1 = cv2.inRange(hsv, red1, red2)
mask_red2 = cv2.inRange(hsv, red3, red4)
mask_red = cv2.bitwise_or(mask_red1, mask_red2)

blueImage = cv2.bitwise_and(image, image, mask=mask_blue)
redImage = cv2.bitwise_and(image, image, mask=mask_red)

color_frame = cv2.hconcat([blueImage, redImage])
cv2_imshow(color_frame)
```

## 데이터 과학
</div>

Colab을 통해 인기 있는 Python 라이브러리를 최대한 활용하여 데이터를 분석하고 시각화할 수 있습니다. 아래 코드 셀에서는 <strong>Numpy</strong>를 사용하여 임의의 데이터를 생성하고 <strong>매트플롯립</strong>으로 이를 시각화합니다. 셀을 클릭하면 코드 수정을 바로 시작할 수 있습니다.

Google Drive 계정에서 스프레드시트를 비롯한 데이터를 Colab 메모장으로 가져오거나 GitHub 등의 여러 다른 소스에서 데이터를 가져올 수 있습니다. Colab을 데이터 과학에 활용하는 방법과 데이터 가져오기에 관해 자세히 알아보려면 <a href="#working-with-data">데이터 사용하기</a> 아래 링크를 참조하세요.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

Colab 노트북은 Google 클라우드 서버에서 코드를 실행하므로 사용 중인 컴퓨터 성능과 관계없이 <a href="#using-accelerated-hardware">GPU 및 TPU</a>를 포함한 Google 하드웨어 성능을 활용할 수 있습니다. 브라우저만 있으면 사용 가능합니다.

예를 들어 <strong>Pandas</strong> 코드의 실행이 완료되기를 기다리고 있는데 더 빠르게 실행하고 싶다면 GPU 런타임으로 전환하고 코드를 변경하지 않고도 가속할 수 있는 <a href="https://rapids.ai/cudf-pandas">RAPIDS cuDF</a>와 같은 라이브러리를 사용할 수 있습니다.

Colab에서 Pandas를 가속하는 방법을 자세히 알아보려면 <a href="https://colab.research.google.com/github/rapidsai-community/showcase/blob/main/getting_started_tutorials/cudf_pandas_colab_demo.ipynb">10분 가이드</a> 또는 <a href="https://colab.research.google.com/github/rapidsai-community/showcase/blob/main/getting_started_tutorials/cudf_pandas_stocks_demo.ipynb">미국 주식 시장 데이터 분석 데모</a>를 참고하세요.
